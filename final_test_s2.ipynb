```python
import pandas as pd

# Example: your dataset
# data2 = pd.read_csv("your_dataset.csv")

# Conversion constants
ML_TO_FLOZ = 0.033814   # 1 mL = 0.033814 fl oz
L_TO_FLOZ = 33.814      # 1 liter = 33.814 fl oz
GRAM_TO_LB = 0.00220462 # 1 gram = 0.00220462 pound
KG_TO_LB = 2.20462      # 1 kg = 2.20462 pound

def normalize_units(row):
    unit = str(row['unit']).lower().strip()
    value = row['value']

    # -----------------------------
    # COUNTABLE UNITS
    # -----------------------------
    if unit in ['count', 'ct', 'each', 'pack', 'packs', 'piece', 'each / pack: 1']:
        return value, 'count'

    # -----------------------------
    # WEIGHT UNITS
    # -----------------------------
    elif unit in ['oz', 'ounce', 'ounces']:
        return value, 'oz'
    elif unit in ['pound', 'lb', 'lbs', 'pounds']:
        return value, 'pound'
    elif unit in ['gram', 'grams', 'gramm', 'g']:
        return value * GRAM_TO_LB, 'pound'
    elif unit in ['kg', 'kilogram', 'kilograms']:
        return value * KG_TO_LB, 'pound'

    # -----------------------------
    # LIQUID UNITS → convert all to fl oz
    # -----------------------------
    elif unit in ['fl oz', 'fl. oz', 'fluid ounce', 'fluid ounces', 'fluid ounce(s)']:
        return value, 'fl oz'
    elif unit in ['ml', 'milliliter', 'milliliters', 'millilitre', 'millilitres', 'mililitro']:
        return value * ML_TO_FLOZ, 'fl oz'
    elif unit in ['l', 'ltr', 'liter', 'liters']:
        return value * L_TO_FLOZ, 'fl oz'

    # -----------------------------
    # DEFAULT: leave as-is (optional)
    # -----------------------------
    else:
        return value, unit

# Apply the function
data2[['value_normalized', 'unit_normalized']] = data2.apply(
    normalize_units, axis=1, result_type='expand'
)

# Optional: check results
print(data2[['unit', 'value', 'unit_normalized', 'value_normalized']].head(20))
```

```python
# Define liquid and count units
liquid_units = ["fl.oz", "fl oz", "fl ounce","fl. oz.","fluid_ounces",]  # normalize to "fl oz"
count_units = [
    "bottle", "can", "bag", "jar", "per carton", "k-cups","tin","sachet","cou",
    "pouch", "tea bags", "paper cupcake liners", "1", "box","unit","2","pac","jars","pieces","count / count","stück","comes as a single dispenser unit with 1 count ...","(pack of 1)"
    "ziplock bags", "carton" , "bottles","bucket","units","8","24","per package","comes as a single 0.1 oz stick for on-the-go use","container","packet","kit","12.54","case / pack: 6"
]
oz=["product_weight","oz.","20 oz.","16.9 oz","1.76 ounce","16 ounces"]

# Function to normalize units
def normalize_unit(x):
    if pd.isna(x):
        return None
    x = str(x).lower().strip()
    
    # Normalize specific liquid units
    if x in liquid_units:
        return "fl oz"
    
    # Map other units to 'count'
    elif x in count_units:
        return "count"

    # Map other units to 'count'
    elif x in oz:
        return "oz"
    
    # Otherwise keep original
    else:
        return x

# Apply normalization
final_data['unit_normalized'] = final_data['unit_normalized'].apply(normalize_unit)

# Optional: check results
print(final_data[['unit_normalized']].head(20))
```

```python
# Count occurrences
norm_unit_counts =final_data["unit_normalized"].value_counts()

# Convert to DataFrame
norm_unit_counts_df = norm_unit_counts.reset_index()
norm_unit_counts_df.columns = ['unit', 'count']

# Show all rows
pd.set_option('display.max_rows', None)
print(norm_unit_counts_df)
```

```python
import torch
import torch.nn as nn

# Map unit to index
unit_encoder = {unit: i for i, unit in enumerate(final_data['unit_normalized'].unique())}

# Create unit_idx in final_data
final_data['unit_idx'] = final_data['unit_normalized'].map(unit_encoder)

# PyTorch embedding layer
num_units = len(unit_encoder)      # 3 in your case
embedding_dim = 4                  # small number since only 3 units
embedding_layer = nn.Embedding(num_units, embedding_dim)

# Convert to tensor
unit_tensor = torch.tensor(final_data['unit_idx'].values, dtype=torch.long)

# Get embeddings
unit_embedded = embedding_layer(unit_tensor)  # shape: [num_rows, embedding_dim]
```
