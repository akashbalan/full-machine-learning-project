```python
import pandas as pd

# Example: your dataset
# data2 = pd.read_csv("your_dataset.csv")

# Conversion constants
ML_TO_FLOZ = 0.033814   # 1 mL = 0.033814 fl oz
L_TO_FLOZ = 33.814      # 1 liter = 33.814 fl oz
GRAM_TO_LB = 0.00220462 # 1 gram = 0.00220462 pound
KG_TO_LB = 2.20462      # 1 kg = 2.20462 pound

def normalize_units(row):
    unit = str(row['unit']).lower().strip()
    value = row['value']

    # -----------------------------
    # COUNTABLE UNITS
    # -----------------------------
    if unit in ['count', 'ct', 'each', 'pack', 'packs', 'piece', 'each / pack: 1']:
        return value, 'count'

    # -----------------------------
    # WEIGHT UNITS
    # -----------------------------
    elif unit in ['oz', 'ounce', 'ounces']:
        return value, 'oz'
    elif unit in ['pound', 'lb', 'lbs', 'pounds']:
        return value, 'pound'
    elif unit in ['gram', 'grams', 'gramm', 'g']:
        return value * GRAM_TO_LB, 'pound'
    elif unit in ['kg', 'kilogram', 'kilograms']:
        return value * KG_TO_LB, 'pound'

    # -----------------------------
    # LIQUID UNITS → convert all to fl oz
    # -----------------------------
    elif unit in ['fl oz', 'fl. oz', 'fluid ounce', 'fluid ounces', 'fluid ounce(s)']:
        return value, 'fl oz'
    elif unit in ['ml', 'milliliter', 'milliliters', 'millilitre', 'millilitres', 'mililitro']:
        return value * ML_TO_FLOZ, 'fl oz'
    elif unit in ['l', 'ltr', 'liter', 'liters']:
        return value * L_TO_FLOZ, 'fl oz'

    # -----------------------------
    # DEFAULT: leave as-is (optional)
    # -----------------------------
    else:
        return value, unit

# Apply the function
data2[['value_normalized', 'unit_normalized']] = data2.apply(
    normalize_units, axis=1, result_type='expand'
)

# Optional: check results
print(data2[['unit', 'value', 'unit_normalized', 'value_normalized']].head(20))
```

```python
# Define liquid and count units
liquid_units = ["fl.oz", "fl oz", "oz.","fl ounce","fl. oz.","20 oz."]  # normalize to "fl oz"
count_units = [
    "bottle", "can", "bag", "jar", "per carton", "k-cups", 
    "pouch", "tea bags", "paper cupcake liners", "1", "box", 
    "ziplock bags", "carton" , "bottles","bucket","units","8","24","per package","comes as a single 0.1 oz stick for on-the-go use"
]

# Function to normalize units
def normalize_unit(x):
    if pd.isna(x):
        return None
    x = str(x).lower().strip()
    
    # Normalize specific liquid units
    if x in liquid_units:
        return "fl oz"
    
    # Map other units to 'count'
    elif x in count_units:
        return "count"
    
    # Otherwise keep original
    else:
        return x

# Apply normalization
final_data['unit_normalized'] = final_data['unit_normalized'].apply(normalize_unit)

# Optional: check results
print(final_data[['unit_normalized']].head(20))
```
